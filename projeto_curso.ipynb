###Covariância e Correlação

|Aspecto|Covariância  .cov()|Correlação  .corr()|
| - | - | - |
|Definição|Mede a direção da relação linear entre duas variáveis|Mede a instensidade e a direção da relação linear entre duas vairáveis|
|intervalos de valores|(-infinto, +infinito)|[-1, +1]|
|Unidade de medida|Depende das unidades das variáveis analisadas|Não possui unidade, pois é uma medida padronizada|
|Interpretação|Valores positivos indicam uma relação direta, negativos indicam uma relação inversa|Valores próximos de 1 ou -1 indicam forte relação linear,enquanto valores próximos de 0 indicam fraca ou nenhuma relação|
|Diferença principal|A covariância não é padronizada, o que dificulta sua interpretação direta|A correlação é a covariância normalizada, facilitando a comparação entre diferentes conjuntos de dados.|

```python
df_number = df_churn.select_dtypes(include=['number', 'bool'])
df_number = df_number.drop(columns='CustomerId', axis=1)
```

```python
cov_matrix = df_number.cov()
cov_matrix
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(cov_matrix, annot=True, fmt='.1f', cmap='RdBu', center=0, linewidths=0.5)
plt.title('Matrix de covariância do dataset')
```

```python
cor_matrix = df_number.corr()
cor_matrix
plt.figure(figsize=(10,6))
sns.heatmap(cor_matrix, annot=True, fmt='.2f', cmap='RdBu', center=0, linewidths=0.5)
plt.title('Matrix de correlação do dataset')
```

###Correlação de Pearson e Spearman

|Característica|Correlação de Pearson|Correlação de Spearman|
| - | - | - |
|Mede relação linear?|sim|sim|
|Mede relação monótona?|não|sim|
|Dados continuos|sim|sim|
|Dados ordinais|não|sim|
|Resistente a outliers|não|sim|
|Quando usar|Quando os dados têm um padrão linear e não possuem outliers|Quando os dados não são lineares ou não possuem outliers|

```python
or_matrix = df_number.corr(method='spearman')
cor_matrix
plt.figure(figsize=(10,6))
sns.heatmap(cor_matrix, annot=True, fmt='.2f', cmap='RdBu', center=0, linewidths=0.5)
plt.title('Matrix de correlação do dataset')
```

```python
df_geo_count = df_full_cities.groupby(by=['Geography'])['Exited'].sum().reset_index(name='Churn')
df_geo_count
```

```python
df_geo_count['Total_Customers'] = df_full_cities.groupby(by=['Geography'])['Exited'].count().values
df_geo_count
```

```python
df_geo_count['Chur_rate'] = df_geo_count['Churn']/df_geo_count['Total_Customers']
df_geo_count
```

```python
#outra forma de conseguir o mesmo resultado -> df_geo_count['Churn_rate_2'] = df_full_cities.groupby(by=['Geography'])['Exited'].mean().values
```

```python
fig = sns.barplot(data=df_geo_count, x='Geography', y='Chur_rate', color='red')
fig.set_title('Taxa de rotatividade por país')
fig.set(xlabel='País', ylabel='Taxa de rotatividade')
fig.set_yticklabels([f'{x*100:.0f}%' for x in fig.get_yticks()])


for index, row in df_geo_count.iterrows():
    plt.text(index, row['Chur_rate'], f'{row["Chur_rate"]*100:.1f}%', color='black', va='bottom', ha='center')
```

```python
df_cities_count = df_full_cities.groupby(by=['Geography', 'County'])['Exited'].sum().reset_index(name='Churn')
df_cities_count
```

```python
df_cities_count['Total_Clientes'] = df_full_cities.groupby(by=['Geography', 'County'])['Exited'].count().values
df_cities_count['Churn_rate'] = df_full_cities.groupby(by=['Geography', 'County'])['Exited'].mean().values
df_cities_count
```

```python
df_cities_count = df_cities_count.sort_values(by=['Churn_rate'], ascending=False).reset_index(drop=True)
fig = sns.barplot(data=df_cities_count, x='Churn_rate', y='County', color='red')
fig.set_title('Taxa de rotatividade por cidade sede do banco')
fig.set(xlabel='Taxa de rotatividade ', ylabel='Sede do banco')
fig.set_xticklabels([f'{x*100:.0f}%' for x in fig.get_xticks()])

for index, line in df_cities_count.iterrows():
    plt.text(line['Churn_rate'], index, f'{line["Churn_rate"]*100:.1f}%', color='white', va='center', ha='right')
    
plt.show()
```

###Erros comum na apresentação dos dados

- Erro -> Analistas tendem a mostrar gráficos sem explicar o que significam
- Acerto -> Gestores querem `recomendações/decisões`. Mostre gráficos com informação do que mostram (call to action), ou seja, coloque um resumo das informações/recomendações adquiridas nos gráficos
- Cuidados: 
    - Execesso de informação -> menos é mais
    - Falta de direção
    - Nenhuma recomendação/informação clara

- Erro -> uso das corres de forma errada
- Acerto -> uso das corres de forma certa, ou seja, verde -> certo; vermelho -> errado; amarelo -> alerta
