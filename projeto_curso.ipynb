###Dicionário de dados

|variável|Descrição|Efeito da saída do cliente|
| - | - | - |
|RowNumber|número da linha no banco de dados|não tem efeito|
|CustomerId|id do cliente|não tem efeito|
|Surname|sobrenome do cliente|não tem efeito|
|CreditScore|Score de crédito do cliente no banco|Provavelmente tem efeito|
|Geography|país de residência do cliente do banco|Provavelmente tem efeito|
|Gender|gênero|pode ter efeito|
|Age|idade do clinete|pode ter efeito|
|Tenure|número de anos de fidelidade do cliente com o banco|Provavelmente tem efeito|
|Balance|saldo do cliente na conta do banco|Provavelmente tem efeito|
|NumOfProducts|número de produtos que o clinete já comprou do banco|pode ter efeito|
|HasCrCard|se o cliente tem cartão de crédito ou não|pode ter efeito|
|IsActiveMember|se é cliente ativo ou não|Provavelmente tem efeito|
|EstimatedSalary|salário estimado do clinete|pode ter efeito|
|Exited|se o cliente saiu ou não do banco|importante|
|Complain|se o cliente registrou algum reclamação ou não|Provavelmente tem efeito|
|Satisfaction Score|nota de satisfação do cliente pela resolução de sua reclamação|pode ter efeito|
|Card Type|tipo de cartão do cliente: pode indicar a classe de fidelidade do cliente|pode ter efeito|
|Point Earned|total de pontos ganhos pelo cliene por uso do cartão de crédito|pode ter efeito|
